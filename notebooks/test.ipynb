```python
import pandas as pd
df = pd.read_csv("../data/processed/feature_engineered_with_SLM.csv")
df.head()
```

```python
print(df.shape[1])
```

```python
for col, dtype in df.dtypes.items():
    print(f"{col}: {dtype}")
```

```python
fraud_count = (df['is_fraud'] == 1).sum()
print(f"Number of fraudulent transactions: {fraud_count}")
```

```python
total_rows = len(df)
fraud_cases = (df['is_fraud'] == 1).sum()
fraud_percentage = (fraud_cases / total_rows) * 100

print(f"Total number of rows: {total_rows}")
print(f"Number of is_fraud with value 1: {fraud_cases}")
print(f"Percentage of is_fraud: {fraud_percentage:.2f}%")
```

```python
import os
import glob

# try expected filename first, else pick the first CSV in ../data/raw
expected = "../data/raw/feature_engineered_with_SLM.csv"
paths = glob.glob("../data/raw/*.csv")
if os.path.exists(expected):
    raw_path = expected
elif paths:
    raw_path = paths[0]
else:
    raise FileNotFoundError("No CSV files found in ../data/raw")

df_raw = pd.read_csv(raw_path)
df_raw.head()

# number of columns
print(df_raw.shape[1])

# dtypes
for c, dt in df_raw.dtypes.items():
    print(f"{c}: {dt}")

# fraud counts and percentage
fraud_count_raw = (df_raw['is_fraud'] == 1).sum()
print(f"Number of fraudulent transactions: {fraud_count_raw}")

total_rows_raw = len(df_raw)
fraud_cases_raw = fraud_count_raw
fraud_percentage_raw = (fraud_cases_raw / total_rows_raw) * 100

print(f"Total number of rows: {total_rows_raw}")
print(f"Number of is_fraud with value 1: {fraud_cases_raw}")
print(f"Percentage of is_fraud: {fraud_percentage_raw:.2f}%")
```
